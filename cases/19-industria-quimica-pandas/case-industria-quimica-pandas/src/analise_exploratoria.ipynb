```python
from pathlib import Path
import pandas as pd
BASE=Path.cwd().parent if Path.cwd().name=='src' else Path.cwd()
vendas=pd.read_csv(BASE/'dados_tratados/fato_vendas.csv')
vendas.info()
```

```python
vendas.groupby('produto_id').agg(receita=('receita_liquida','sum'), margem=('margem','sum'), pedidos=('pedido_id','nunique')).sort_values('receita',ascending=False)
```

```python
vendas.assign(competencia=pd.to_datetime(vendas.data_pedido).dt.to_period('M')).groupby('competencia')['receita_liquida'].sum().plot(figsize=(12,4),title='Receita mensal')
```
